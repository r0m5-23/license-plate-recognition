# **Bibliotecas**

```python
from paddleocr import PaddleOCR, draw_ocr # paddleOCR
from google.colab.patches import cv2_imshow # visualização de imagens
from PIL import Image # para ler e guardar as imagens
import glob # Para obter o caminho de ficheiros presentes em diretorias
```

# **Carregar modelo OCR**

```python
paddle_ocr = PaddleOCR(use_angle_cls=True, lang='pt', use_gpu=False)
```

```python
from IPython.display import display

imagens = glob.glob("/content/drive/MyDrive/PDI/Matriculas/" + "*.png")
caminho_ficheiro_resultados = "/content/drive/MyDrive/PDI/Matriculas/resultados.txt" # caminho onde será guardado o ficheiro

for item in imagens:

  result = paddle_ocr.ocr(item)

  if result is not None and result[0] is not None:
    boxes = [line[0][0] for line in result]
    txts = [line[0][1][0] for line in result]
    scores = [line[0][1][1] for line in result]
  else:
    txts = [0]
    scores = [0]

  #show

  im_show = draw_ocr(Image.open(item).convert('RGB'), boxes, txts, scores, font_path='/content/PaddleOCR/StyleText/fonts/en_standard.ttf') # desenhar a bounding box na imagem original e adicionar o resultado e a sua confiança

  im_show = Image.fromarray(im_show) # converter um array para uma imagem

  display(im_show) # visualização do resultado

  # gera o ficheiro caso não exista
  with open(caminho_ficheiro_resultados, "a") as resultados_f:
      for text, confidence in zip(txts, scores):
        resultados_f.write(f"{text}, {confidence:.2f}\n")


```
