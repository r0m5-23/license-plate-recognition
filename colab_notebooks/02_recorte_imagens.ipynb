```python
import cv2 # OpenCV
import glob # Glob
import os

# visualização de imagens
from google.colab.patches import cv2_imshow
```

# **Ler imagens**



```python

imagens = glob.glob("/content/drive/MyDrive/PDI/yolov5-master/runs/detect/resultados_inferencia20/" + "*.png") # os caminhos das imagens 13

nomes_imagens = []

for item in imagens: # correr cada imagem

  nome_imagem = item.split("/")[-1].replace(".png", "")

  nomes_imagens.append(nome_imagem)
```

# **Ler ficheiros de texto**

```python

labels = glob.glob("/content/drive/MyDrive/PDI/yolov5-master/runs/detect/resultados_inferencia20/labels/" + "*.txt") # os caminhos dos ficheiros 13

nomes_labels = []

for item in labels: # correr cada label

  nome_label = item.split("/")[-1].replace(".txt", "")

  nomes_labels.append(nome_label)
```

# **Recortar a imagem**

```python
# imagem.shape -> array com as dimensões da imagem
# shape[0] -> largura
# shape[1] -> altura

save_dir = "/content/drive/MyDrive/PDI/Matriculas/"
os.makedirs(save_dir, exist_ok=True)  # Criar diretoria se ela não existir

counter = 0

for i in range (0, len(nomes_imagens)):

  if nomes_imagens[i] in nomes_labels:

    caminho_imagem = "/content/drive/MyDrive/PDI/yolov5-master/runs/detect/resultados_inferencia20/" + nomes_imagens[i] + ".png"
    caminho_label = "/content/drive/MyDrive/PDI/yolov5-master/runs/detect/resultados_inferencia20/labels/" + nomes_imagens[i] + ".txt"

    imagem = cv2.imread(caminho_imagem) # ler a imagem

    largura = imagem.shape[1] # largura da imagem
    altura = imagem.shape[0] # altura da imagem

    f = open(caminho_label)

    linhas = f.readlines() # ler todas as linhas presentes no ficheiro

    # visualizar todas as linhas do ficheiro
    for l in linhas:

      l = l.replace("\n", "") # remover o parágrafo presente no final de cada linha

      l = l.split(" ") # obter cada valor de forma individualizada

      coordenadas_normalizadas = [float(l[1]), float(l[2]), float(l[3]), float(l[4])]

      # reverter a normalização das coordenadas

      des_x = coordenadas_normalizadas[0] * largura
      des_y = coordenadas_normalizadas[1] * altura
      des_width = coordenadas_normalizadas[2] * largura
      des_height = coordenadas_normalizadas[3] * altura

      # calcular os pontos

      bb_espessura = 1 # para remover a bounding box

      xmin = des_x - des_width + des_width / 2 + bb_espessura
      xmax = des_x + des_width - des_width / 2 - bb_espessura
      ymin = des_y - des_height + des_height / 2 + bb_espessura
      ymax = des_y + des_height - des_height / 2 - bb_espessura

      # recorte da imagem -> os valores têm de ser inteiros
      imagem_crop = imagem[int(ymin) : int(ymax), int(xmin) : int(xmax)]

      # redimensionar a imagem
      imagem_crop = cv2.resize(imagem_crop, (300, 75))

      cv2_imshow(imagem_crop) # visualizar a imagem

      # guardar imagem
      cv2.imwrite(os.path.join(save_dir, f"carro{counter}.png"), imagem_crop)

      counter = counter + 1

    # fechar o ficheiro
    f.close()

print(f"{counter} matriculas saved")

```
