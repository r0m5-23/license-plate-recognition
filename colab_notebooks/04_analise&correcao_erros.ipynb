## **Análise de texto e correção de erros**

```python
caminho_ficheiro_resultados = "/content/drive/MyDrive/PDI/Matriculas/resultados.txt"
caminho_ficheiro_resultados_corrigidos = "/content/drive/MyDrive/PDI/Matriculas/resultados_corrigidos.txt"

with open(caminho_ficheiro_resultados, "r") as resultados_f:
  linhas = resultados_f.readlines()

with open(caminho_ficheiro_resultados_corrigidos, "w") as resultados_corrigidos_f:
  for linha in linhas:
    linha_dividida = linha.split(", ")
    matricula, confianca = linha_dividida

    matricula_corrigida = ''

    for c in matricula:

      if c.isalnum():
        matricula_corrigida += c

    if len(matricula_corrigida) == 6 and any(c.isalpha() for c in matricula_corrigida) and any(c.isdigit() for c in matricula_corrigida):

      letras = sum(c.isalpha() for c in matricula_corrigida)
      numeros = sum(c.isdigit() for c in matricula_corrigida)

      for i in range(0, len(matricula_corrigida)-1, 2):

        par = matricula_corrigida[i:i+2]

        if letras > 4 and numeros < 2:

          if par[0].isalpha() and par[1].isdigit():
            par = par[0].replace('O', '0').replace('I', '1') + par[1]
          elif par[0].isdigit() and par[1].isalpha():
            par = par[0] + par[1].replace('O', '0').replace('I', '1')
          elif par[0].isalpha() and par[1].isalpha():
            par = par[0].replace('O', '0').replace('I', '1') + par[1].replace('O', '0').replace('I', '1')

        elif numeros > 4 and letras < 2:

          if par[0].isalpha() and par[1].isdigit():
            par = par[0] + par[1].replace('0', 'O').replace('1', 'I')
          elif par[0].isdigit() and par[1].isalpha():
            par = par[0].replace('0', 'O').replace('1', 'I') + par[1]
          elif par[0].isdigit() and par[1].isdigit():
            par = par[0].replace('0', 'O').replace('1', 'I') + par[1].replace('0', 'O').replace('1', 'I')

        elif 2 < letras < 4 and 2 < numeros < 4:

          if par[0].isalpha() and par[1].isdigit():
            par = par[0] + par[1].replace('0', 'O').replace('1', 'I')
          elif par[0].isdigit() and par[1].isalpha():
            par = par[0].replace('0', 'O').replace('1', 'I') + par[1]

        matricula_corrigida = matricula_corrigida[:i] + par + matricula_corrigida[i+2:]

      resultados_corrigidos_f.write(f"{matricula_corrigida}\n")

```
